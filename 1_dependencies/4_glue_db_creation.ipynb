# Database Creation

This document will be responsible for setting up the databases that will house the data using AWS Glue Catalog

```python
# import libraries
import boto3

from botocore.exceptions import ClientError
```

```python
# initialize glue client
glue = boto3.client('glue')
```

```python
# define raw_data db as object
raw_db_name = 'raw_data'
raw_db_inputs = {
    'Name': raw_db_name,
    'Description': 'Databse for raw data, two tables one for flight and one for weather'
}
```

```python
# create raw_data Glue db
response = glue.create_database(DatabaseInput=raw_db_inputs)
```

```python
if response['ResponseMetadata']['HTTPStatusCode'] == 200:
    print(f"Database '{raw_db_name}' creation request succeeded.")

    try:
        db_info = glue.get_database(Name=raw_db_name)
        print(f"Successfully retrieved database info: {db_info['Database']['Name']}")
    except ClientError as e:
        print(f"Error retrieving databse '{raw_db_name}':", e)
else:
    print("Error creating databse. Response:", response)
```

```python
# define processed_data db as object
proc_db_name = 'processed_data'
processed_db_inputs = {
    'Name': proc_db_name,
    'Description': 'Database for processed data with one table where each flight has weather information predictors'
}
```

```python
# create processed_data db
response1 = glue.create_database(DatabaseInput=processed_db_inputs)
```

```python
if response1['ResponseMetadata']['HTTPStatusCode'] == 200:
    print(f"Database '{proc_db_name}' creation request succeeded.")

    try:
        db_info = glue.get_database(Name=proc_db_name)
        print(f"Successfully retrieved database info: {db_info['Database']['Name']}")
    except ClientError as e:
        print(f"Error retrieving databse '{proc_db_name}':", e)
else:
    print("Error creating databse. Response:", response1)
```
