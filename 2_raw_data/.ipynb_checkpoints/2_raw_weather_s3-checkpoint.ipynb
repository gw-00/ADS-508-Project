# Raw Weather S3 Bucket Upload

```python
# import libraries
import pandas as pd
import boto3
import os
```

```python
# read in CSV's to pandas dataframes
df1 = pd.read_csv('raw_weather_data/KSAN_2023_wx.csv')
df2 = pd.read_csv('raw_weather_data/KSAN_2024_wx.csv')
df3 = pd.read_csv('raw_weather_data/KSAN_2025_wx.csv')
```

```python
# create a list of the data frames
dfs = [df1, df2, df3]
```

```python
# merge the dataframes using concat
merged_dfs = pd.concat(dfs, ignore_index=True)
```

```python
# examine the head of the df
merged_dfs.head()
```

```python
# examine the tail of the df
merged_dfs.tail()
```

```python
# save the merged df to csv
merged_dfs.to_csv('raw_weather_data/merged_wx.csv', index=False)
```

```python
# examine list of buckets available
s3 = boto3.client('s3')
response = s3.list_buckets()
print("Existing buckets:")
for bucket in response['Buckets']:
    print(bucket['Name'])
```

```python
# set local path of merged CSV file
local_file_path = 'raw_weather_data/merged_wx.csv'
```

```python
# set the raw data bucket variable
raw_bucket = "group9-ml-project-raw-data-bucket-grw"
```

```python
# set the S3 key path, we are using flat structure because this will only hold two files
s3_key = "data/merged_wx.csv"
```

```python
# upload file to the S3 bucket
try:
    s3.upload_file(local_file_path, raw_bucket, s3_key)
    print(f"Uploaded {local_file_path} to s3://{raw_bucket}/{s3_key}")
except Exception as e:
    print("Error uploading file to S3:", e)
```
